### Second training run for person dataset
Changes:
- **Batch** was upped from **16** to **64** to increase training times since we could afford to do so with the A100.
- Some hyperparameters were changed like:
    - Learning Rate Final (lrf) increased from 0.01 to 0.2 to slow learning rate decay to improve generalisation
    - Object loss weight (obj) increased from 1.0 to 1.2 to attempt to improve sensitivity of occluded images
    - Mixup was upped from 0 to 0.1 to allow for scene blending, attempting to improve generalisation.
    - Updated model weights to start from the best weight of the previous run.

```python
# Switch to YOLOv5 directory
%cd /content/yolov5

# Train with tuned hyperparams and better efficiency
!python train.py \
  --img 640 \
  --batch 64 \
  --epochs 20 \
  --data /content/datasets/coco-person/data.yaml \
  --cfg models/yolov5s.yaml \
  --weights /content/drive/MyDrive/yolov5_runs_backup/person_detector_yolov5s_20epoch_local/weights/best.pt \
  --hyp /content/hyp_v2.yaml \
  --workers 4 \
  --cache ram \
  --name person_detector_yolov5s_20epoch_local_v2

# Example: copy all training results to Drive
!cp -r /content/yolov5/runs/train /content/drive/MyDrive/yolov5_runs_backup/
```

After the second run we don't see too many massive differences in improvements, however there is still some improvement and we are happy to move on to threat detection with the same settings and hyperparameters to the threat detection phase.

```python
# Preparing hyperparameters for threat detection, these are the hyperparameters I used for training the threat detection model
hyp_threat_yaml = """
lr0: 0.01
lrf: 0.2
momentum: 0.937
weight_decay: 0.0005
warmup_epochs: 3.0
warmup_momentum: 0.8
warmup_bias_lr: 0.1
box: 0.05
cls: 0.5
cls_pw: 1.0
obj: 1.3
obj_pw: 1.0
iou_t: 0.2
anchor_t: 4.0
fl_gamma: 0.0
hsv_h: 0.015
hsv_s: 0.7
hsv_v: 0.4
degrees: 0.0
translate: 0.1
scale: 0.6
shear: 0.0
perspective: 0.0
flipud: 0.0
fliplr: 0.5
mosaic: 1.0
mixup: 0.1
copy_paste: 0.0
"""

with open("/content/hyp_threat.yaml", "w") as f:
    f.write(hyp_threat_yaml.strip())

print("✅ hyp_threat.yaml written to /content/hyp_threat.yaml")
```

```python
!cat /content/hyp_threat.yaml
```

```python
# Switch to YOLOv5 directory
%cd /content/yolov5

# Train the final combined detector with tuned hyperparameters and correct dataset
!python train.py \
  --img 512 \
  --batch 32 \
  --epochs 20 \
  --data /content/final-combined-dataset/data.yaml \
  --cfg models/yolov5s.yaml \
  --weights yolov5s.pt \
  --hyp /content/hyp_threat.yaml \
  --workers 4 \
  --cache \
  --name final_combined_detector_yolov5s_30epoch_local_v1

  # Example: copy all training results to Drive
!cp -r /content/yolov5/runs/train /content/drive/MyDrive/yolov5_runs_backup_v2/
```
