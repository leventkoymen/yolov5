## Dataset Download

```python
import fiftyone.zoo as foz

# List of threat detection classes
threat_classes = [
    "Backpack", "Baseball bat", "Bow and arrow",
    "Dagger", "Handgun", "Knife", "Missile", "Rifle",
    "Scissors", "Shotgun", "Suitcase", "Sword", "Truck", "Weapon"
]

# Download full dataset with splits
# Note: This will create and store the dataset in the default FiftyOne dataset directory on your computer, not here
dataset = foz.load_zoo_dataset(
    "open-images-v7",
    splits=["train", "validation", "test"],
    classes=threat_classes,
    label_types=["detections"],
    shuffle=True,
    seed=51,
    dataset_name="open-images-threat-detection"
)
```

## Open Images V7 Verification Code
This code simply verifies that it has downloaded and displays the per split sample count but also per class sample count as well.

```python
import fiftyone as fo
import collections

dataset = fo.load_dataset("open-images-threat-detection")

# Prints the number of samples in each split
for split in ["train", "validation", "test"]:
    view = dataset.match_tags(split)
    print(f"{split}: {len(view)} samples")

# Your threat classes
classes = [
    "Backpack", "Baseball bat", "Bow and arrow",
    "Dagger", "Handgun", "Knife", "Missile", "Rifle",
    "Scissors", "Shotgun", "Suitcase", "Sword", "Truck", "Weapon"
]

# Initialize counters for each split
split_class_counts = {
    "train": collections.Counter(),
    "validation": collections.Counter(),
    "test": collections.Counter(),
}


for sample in dataset:
    split = None
    if "train" in sample.tags:
        split = "train"
    elif "validation" in sample.tags:
        split = "validation"
    elif "test" in sample.tags:
        split = "test"
    
    if split is None:
        continue

    # Count labels within the sample
    detections = sample.ground_truth.detections
    for det in detections:
        if det.label in classes:
            split_class_counts[split][det.label] += 1

# Displays the counts of each class in each split in an easy-to-read format
for cls in classes:
    train_count = split_class_counts["train"].get(cls, 0)
    val_count = split_class_counts["validation"].get(cls, 0)
    test_count = split_class_counts["test"].get(cls, 0)
    print(f"{cls}: Train={train_count}, Val={val_count}, Test={test_count}")
```
