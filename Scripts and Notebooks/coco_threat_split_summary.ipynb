# COCO Threat Split Summary

```python
import pandas as pd
from pycocotools.coco import COCO

train_path = "../datasets/coco/annotations/instances_train2017.json"
val_path = "../datasets/coco/annotations/instances_val2017.json"

threat_classes = [
    "truck", "backpack", "suitcase", "baseball bat", "knife", "scissors"
]

def get_split_counts(annotation_path, class_names):
    coco = COCO(annotation_path)
    class_id_to_name = {cat['id']: cat['name'] for cat in coco.loadCats(coco.getCatIds())}
    name_to_class_id = {v: k for k, v in class_id_to_name.items()}
    
    summary = {}
    for cls in class_names:
        cat_id = name_to_class_id.get(cls)
        if cat_id is not None:
            img_ids = coco.getImgIds(catIds=cat_id)
            summary[cls] = len(set(img_ids))
        else:
            summary[cls] = 0
    return summary

train_counts = get_split_counts(train_path, threat_classes)
val_counts = get_split_counts(val_path, threat_classes)

df = pd.DataFrame({
    "Class": threat_classes,
    "Train": [train_counts[c] for c in threat_classes],
    "Validation": [val_counts[c] for c in threat_classes]
})
print(df)
```
